```python
nyc_inv[nyc_inv["Roadway_Name"].str.contains("EXPY")][["Roadway_Name", "Roadway_Type","Owning_Muni_Name", "Owning_Jurisdiction"]]
```

```python
#exp only

exp_inv = nyc_inv[nyc_inv["Roadway_Name"].str.contains("EXPRESSWAY|EXPY|EXP")]
exp_inv.shape
```

```python
exp_inv["County_Name"].unique()
```

```python
CBE_inv = exp_inv[exp_inv["Roadway_Name"].str.contains("LONG ISLAND")][["OBJECTID","Route_ID","Roadway_Name", "Owning_Muni_Name"]]
CBE_inv
```
